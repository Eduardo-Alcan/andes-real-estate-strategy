# Análisis Comercial Inmobiliario — Grupo Andes

## Resumen ejecutivo

Este proyecto evalúa el desempeño comercial de Grupo Andes durante el periodo analizado, con el objetivo de identificar qué segmentos del negocio generan más valor y dónde existen oportunidades de crecimiento. Elegí este proyecto porque me permitió trabajar el ciclo completo de un análisis inmobiliario: desde el modelado de datos hasta la construcción de un dashboard ejecutivo en Power BI.

**Hallazgos principales:**

- El negocio generó un ingreso total de **$6.01 mil millones** a partir de **8,500 ventas**, con un ticket promedio de **$707 mil** y una comisión total de **$200.6 millones**.
- **Casa** es el tipo de propiedad que más ingresos genera, superando los **$2.2 mil millones**.
- **Bogotá** concentra el mayor volumen de ventas, con **4,377 transacciones**.
- El canal **Corredor** domina la operación comercial, representando aproximadamente el **73%** del volumen de ventas.
- El segmento **"Primera vez"** concentra la mayor parte del ingreso, y las cohortes de clientes adquiridos en **2023-03 / 2023-04** muestran mayor recurrencia e ingresos acumulados que las cohortes más recientes.

**Recomendaciones estratégicas:**

1. Priorizar la comercialización de propiedades tipo Casa, dada su mayor contribución al ingreso total.
2. Fortalecer el canal Directo para diversificar la captación comercial y reducir la dependencia del canal Corredor.
3. Implementar estrategias de retención enfocadas en cohortes recientes, replicando lo que funcionó en las cohortes de 2023-03/04.
4. Diseñar campañas específicas para clientes de primera compra, dado su peso dentro del ingreso total del negocio.

El detalle metodológico, el modelo de datos y el diseño del dashboard se documentan a continuación.

## Contexto y objetivo

Grupo Andes necesitaba una forma clara de monitorear su desempeño comercial: qué tan rentable es cada tipo de propiedad, qué canales de venta son más eficientes y si los clientes vuelven a comprar después de su primera transacción. Antes de este análisis, esa información vivía dispersa entre transacciones, catálogo de propiedades y datos de clientes, sin una vista consolidada.

El objetivo fue construir un dashboard ejecutivo en Power BI que consolidara estas fuentes en un modelo de datos único, y que permitiera a la dirección comercial tomar decisiones basadas en evidencia, no en percepciones sobre qué canal o tipo de propiedad "funciona mejor".

## Fuentes de datos

| Tabla | Descripción |
|---|---|
| `hecho_ventas_propiedades` | Tabla de hechos con las transacciones de venta: precio, cliente, propiedad, canal y fecha |
| `dim_clientes` | Información y segmentación de clientes |
| `dim_propiedades` | Características de las propiedades: tipo, tamaño, ubicación |
| `dim_fecha` | Dimensión de calendario construida para habilitar inteligencia de tiempo (YTD, MTD, YoY) |

El modelo se estructuró en esquema de estrella, con `hecho_ventas_propiedades` como tabla de hechos conectada a las tres dimensiones.

## Preparación y calidad de datos

Antes de construir cualquier métrica, validé la calidad de la información:

- **Tipos de datos:** confirmé que `fecha_venta` estuviera en formato *Date* y que `porcentaje_comision` estuviera correctamente formateado como porcentaje, ya que un error aquí afecta directamente el cálculo de la comisión total.
- **Valores nulos:** revisé la presencia de nulos en cada tabla y evalué caso por caso si debían eliminarse o mantenerse, en lugar de aplicar una regla genérica de "eliminar todo lo vacío".
- **Duplicados:** validé que las claves primarias de `dim_clientes` y `dim_propiedades` no tuvieran registros duplicados, lo que habría inflado artificialmente los conteos de ventas y clientes únicos.

## Modelado temporal

Construí una dimensión de fechas (`dim_fecha`) independiente de las tablas transaccionales, lo que permitió calcular métricas de inteligencia de tiempo (YTD, MTD, comparaciones año contra año) sin depender de que cada fecha existiera necesariamente en la tabla de hechos.

## Métricas de negocio

Definí las métricas en cuatro grupos, según el tipo de pregunta de negocio que resuelven:

**Medidas base**
- Ingreso Total
- Total de Ventas
- Ticket Promedio
- Comisión Total

**Medidas de participación** (usadas como contexto en gráficos y tooltips)
- % de ingresos por tipo de propiedad
- % de ingresos por canal de venta
- % de ingresos por segmento de cliente

**Inteligencia de tiempo**
- Ventas Year to Date (YTD) y Month to Date (MTD)
- Ventas del año anterior
- Crecimiento Year over Year (YoY)

**Análisis de cohortes**
- Mes Cohorte: mes de la primera compra de cada cliente
- Mes Venta: mes de cada transacción posterior
- Matriz de cohortes (Mes Cohorte × Mes Venta) para identificar patrones de recompra

## Diseño del dashboard

El dashboard se organizó en tres vistas, cada una respondiendo un nivel distinto de la pregunta de negocio:

**Overview ejecutivo** — Tarjetas de Ingreso Total, Total de Ventas, Ticket Promedio, Comisión Total y Crecimiento YoY; evolución de ingresos en el tiempo; ingresos por ciudad; filtros por año-mes, ciudad y canal de venta.

**Análisis comercial** — Tabla comparativa por tipo de propiedad (ventas, ingreso, ticket promedio, comisión, crecimiento YoY) y participación de ingresos por tipo de propiedad, canal de venta y segmento de cliente.

**Análisis de cohortes** — Matriz de cohortes para evaluar si los clientes vuelven a comprar después de su primera transacción y qué cohortes generan más ingresos acumulados en el tiempo.

## Insights adicionales

- El segmento de clientes "Primera vez" concentra la mayor parte del ingreso, lo que sugiere que la adquisición de nuevos clientes sigue siendo el principal motor de ventas — y a la vez, una oportunidad de crecimiento si se mejora la retención.
- Las cohortes de 2023-03 y 2023-04 muestran mayor recurrencia e ingresos acumulados que cohortes más recientes, lo que puede indicar un cambio en la estrategia de adquisición o en la calidad del seguimiento post-venta entre esos periodos y los más recientes. Vale la pena investigar qué se hizo distinto en esas cohortes para poder replicarlo.
- La alta concentración del canal Corredor (73% del volumen) representa tanto una fortaleza operativa como un riesgo de dependencia: si ese canal se ve afectado, el impacto en ventas sería significativo.

## Conclusiones

Este análisis muestra que el negocio de Grupo Andes está concentrado en pocos ejes clave: un tipo de propiedad (Casa), una ciudad (Bogotá) y un canal (Corredor) explican la mayor parte del desempeño comercial. Esa concentración ha sido efectiva hasta ahora, pero también expone al negocio a riesgo si alguno de esos ejes se debilita.

Las recomendaciones de este proyecto —diversificar canales, fortalecer la retención de cohortes recientes y capitalizar el peso del segmento de primera compra— buscan mantener el desempeño actual mientras se reduce esa concentración de riesgo.
